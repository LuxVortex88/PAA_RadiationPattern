```python
import numpy as np
import matplotlib.pyplot as plt

# Parameters
N = 8 # Numbers of elements in array
d = 0.5 # Element spacing (in wavelengths)
theta = np.linspace(-np.pi/2, np.pi/2, 180) # Angle range (90° to 90°)

# Array configuration(uniform linear array)
def phase_array_steering(theta, N, d):
    # Assume isotropic elements with a uniform phase shift
    k = 2 * np.pi # Wave Number (assuming wavelength = 1)
    psi = k * d * np.sin(theta)
    return np.abs(np.sin(N * psi / 2) / np.sin(psi / 2))

# Calculate the beamforming pattern
radiation_pattern = phase_array_steering(theta, N, d)

# Plot the radiation pattern
plt.plot(np.degrees(theta), radiation_pattern)
plt.title('Phased Array Antenna Radiation Pattern')
plt.xlabel('Angle (degrees)')
plt.ylabel('Radiation Pattern')
plt.grid(True)
plt.show()
```
